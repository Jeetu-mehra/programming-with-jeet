### **Theory/Background:**
**Apache Hadoop** is an open-source distributed storage and processing framework designed to handle massive amounts of structured and unstructured data across commodity hardware clusters.

The main components of the Apache Hadoop ecosystem include:
- **HDFS (Hadoop Distributed File System):** A distributed, scalable, and portable file system designed with a master-slave architecture (**NameNode** and **DataNodes**) that stores data across nodes in blocks with configurable replication for high fault tolerance.
- **MapReduce:** A software framework and programming model for processing vast amounts of data in parallel across Hadoop clusters.
- **YARN (Yet Another Resource Negotiator):** The cluster management layer that schedules jobs and allocates system compute resources efficiently.
- **Hadoop Common:** The fundamental utilities, libraries, and Java classes required by all Hadoop submodules.

In this experiment, we configure a single-node pseudo-distributed Hadoop environment and execute fundamental HDFS file system management commands.

## **Step 6: Create User Directory in HDFS and Verify**
We create a base working directory for the user in HDFS using `hdfs dfs -mkdir -p` and list the parent directory using `hdfs dfs -ls`.


```python
# Create directory in HDFS and verify
!hdfs dfs -mkdir -p /user/$USER/hadoop_exp5
!hdfs dfs -ls /user/$USER
```

## **Step 7: Create a Local Sample File and Upload to HDFS**
We create a local text file `sample.txt`, inspect its content, upload it to the HDFS directory using `hdfs dfs -put`, and confirm its presence using `hdfs dfs -ls`.


```python
# Create local sample file and upload to HDFS
!echo -e "Hadoop HDFS Experiment 5\nNameNode manages metadata\nDataNode stores data blocks\nHDFS provides distributed storage" > ~/sample.txt
!cat ~/sample.txt
!hdfs dfs -put ~/sample.txt /user/$USER/hadoop_exp5/
!hdfs dfs -ls /user/$USER/hadoop_exp5
```

## **Step 8: Display File Contents from HDFS and Copy within HDFS**
We read the content of the file directly from HDFS using `hdfs dfs -cat` and create a copy inside the HDFS directory using `hdfs dfs -cp`.


```python
# Display file content and copy file within HDFS
!hdfs dfs -cat /user/$USER/hadoop_exp5/sample.txt
!hdfs dfs -cp /user/$USER/hadoop_exp5/sample.txt /user/$USER/hadoop_exp5/sample_copy.txt
!hdfs dfs -ls /user/$USER/hadoop_exp5
```

## **Step 9: Download File from HDFS to Local File System and Verify**
We use `hdfs dfs -get` to download `sample_copy.txt` from HDFS to the local user directory as `sample_download.txt` and verify its contents.


```python
# Download file from HDFS and display local file content
!hdfs dfs -get /user/$USER/hadoop_exp5/sample_copy.txt ~/sample_download.txt
!cat ~/sample_download.txt
```

## **Step 10: Check HDFS Disk Space and Delete File from HDFS**
We check overall HDFS storage capacity using `hdfs dfs -df -h` and delete `sample_copy.txt` from HDFS using `hdfs dfs -rm`.


```python
# Check HDFS disk free space and remove file
!hdfs dfs -df -h
!hdfs dfs -rm /user/$USER/hadoop_exp5/sample_copy.txt
!hdfs dfs -ls /user/$USER/hadoop_exp5
```
